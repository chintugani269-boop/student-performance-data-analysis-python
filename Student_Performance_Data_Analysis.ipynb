```python
df.loc[2, "Math"] = np.nan
df.loc[5, "Science"] = np.nan

df.isnull().sum()
```

```python
df["Math"] = df["Math"].fillna(df["Math"].mean())

df["Science"] = df["Science"].fillna(
    df["Science"].mean()
)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df["Total_Marks"] = (
    df["Math"] +
    df["Science"] +
    df["English"]
)

df.head()
```

```python
df["Total_Marks"]
```

```python
df["Math"] + df["Science"] + df["English"]
```

```python
df["Average_Marks"] = (
    df["Total_Marks"] / 3
)

df["Average_Marks"] = df["Average_Marks"].round(2)

df.head()
```

```python
def performance_category(mark):
    if mark >= 90:
        return "Excellent"
    elif mark >= 75:
        return "Good"
    elif mark >= 60:
        return "Average"
    else:
        return "Poor"


df["Performance"] = df["Average_Marks"].apply(
    performance_category
)

df[["Name", "Average_Marks", "Performance"]].head(10)
```

```python
top_students = df.sort_values(
    by="Average_Marks",
    ascending=False
)

top_students[
    ["Name", "Average_Marks", "Performance"]
].head(5)
```

```python
lowest_students = df.sort_values(
    by="Average_Marks",
    ascending=True
)

lowest_students[
    ["Name", "Average_Marks", "Performance"]
].head(5)
```

```python
high_attendance = df[df["Attendance"] >= 90]

print("Students with 90% or more attendance:")
print(high_attendance[[
    "Name",
    "Attendance",
    "Average_Marks"
]])
```

```python
low_attendance = df[df["Attendance"] < 75]

print("Students with less than 75% attendance:")

if len(low_attendance) > 0:
    print(low_attendance[[
        "Name",
        "Attendance",
        "Average_Marks"
    ]])
else:
    print("No students found with attendance below 75%.")
```

```python
high_study = df[df["Study_Hours"] > 5]

print("Students studying more than 5 hours:")

print(high_study[[
    "Name",
    "Study_Hours",
    "Average_Marks"
]])
```

```python
low_study = df[df["Study_Hours"] <= 5]

print("Students studying 5 hours or less:")

print(low_study[[
    "Name",
    "Study_Hours",
    "Average_Marks"
]])
```

```python
gender_average = df.groupby(
    "Gender"
)["Average_Marks"].mean()

print("Gender-wise Average Marks:")

print(gender_average.round(2))
```

```python
gender_attendance = df.groupby(
    "Gender"
)["Attendance"].mean()

print("Gender-wise Average Attendance:")

print(gender_attendance.round(2))
```

```python
subject_average = df[
    ["Math", "Science", "English"]
].mean()

print("Subject-wise Average Marks:")
print(subject_average.round(2))
```

```python
study_analysis = df.groupby(
    "Study_Hours"
)["Average_Marks"].mean()

print("Study Hours vs Average Marks:")
print(study_analysis.round(2))
```

```python
attendance_analysis = df.groupby(
    "Attendance"
)["Average_Marks"].mean()

print("Attendance vs Average Marks:")
print(attendance_analysis.round(2))
```

```python
correlation = df[
    [
        "Study_Hours",
        "Attendance",
        "Math",
        "Science",
        "English",
        "Average_Marks"
    ]
].corr()

print("Correlation Matrix:")
print(correlation.round(2))
```
